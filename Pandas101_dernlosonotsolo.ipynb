## HW (ให้ใช้เฉพาะที่อาจารย์สอนไปแล้วในวิชานี้)

1.   คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของข้อมูลทั้งหมด
2.   คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของผู้ป่วยในจังหวัดขอนแก่น

````python
this_data = data_covid[['sex','age','province_of_onset']]
````

````python
this_data
````

````python
female = this_data[this_data['sex']=='หญิง']
female['age']
````

````python
female_age_numeric = pd.to_numeric(female['age'], errors='coerce').dropna()

sum_age = 0
N_age = 0
for a in female_age_numeric:
    if a > 0:
        sum_age += a # sum = sum + a
        N_age += 1

if N_age > 0:
    print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง {sum_age/N_age}')
else:
    print('ไม่พบข้อมูลอายุที่เป็นบวกสำหรับผู้ป่วยหญิง')
````

````python
female_KK = female[female['province_of_onset']=='ขอนแก่น']

female_KK_age_numeric = pd.to_numeric(female_KK['age'], errors='coerce').dropna()

sum_age_KK = 0
N_age_KK = 0
for a in female_KK_age_numeric:

    if a > 0:
        sum_age_KK += a
        N_age_KK += 1

if N_age_KK > 0:
    print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง ในจังหวัดขอนแก่น {sum_age_KK/N_age_KK}')
else:
    print('ไม่พบข้อมูลอายุที่เป็นบวกสำหรับผู้ป่วยหญิงในจังหวัดขอนแก่น')
````

````python
male = this_data[this_data['sex']=='ชาย']
male['age']
````

````python
male_age_numeric = pd.to_numeric(male['age'], errors='coerce').dropna()

sum_age = 0
N_age = 0
for a in male_age_numeric:
    if a > 0:
        sum_age += a
        N_age += 1

if N_age > 0:
    print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย {sum_age/N_age}')
else:
    print('ไม่พบข้อมูลอายุที่เป็นบวกสำหรับผู้ป่วยชาย')
````

### การวนลูป record ในตาราง  (```.iterrows```)

````python
this_data = data_covid[['sex','age','province_of_onset']]
````

````python
for each_row in this_data.iterrows():
    # print(each_row)
    # print(each_row[1])
    if (each_row[1]['age'] == 20) and (each_row[1]['province_of_onset'] == 'บุรีรัมย์'):
        print(each_row)
````

### การวนลูป แบบมองตารางแพนด้าส์(pandas dataframe) เป็น numpy array  (```.iloc```)

````python
for each_row in range(this_data.shape[0]): # each_row วนตามแถว this_data.shape[0] ต้องดูตั้งแต่ row ที่ 0
    if (this_data.iloc[each_row,1] == 20) and (this_data.iloc[each_row,2] == 'บุรีรัมย์'):  #each_row คือ ตัวเลข (ตรง 1 คือหลักที่ 2 เป็น 1 งงมั้ยกูงง)
        print(each_row)
        print(this_data.iloc[each_row,:])
````

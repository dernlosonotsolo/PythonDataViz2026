````python
time1 = '12:30:15'

hour, minute, second = time1.split(":")


hour = int(hour)
minute = int(minute)
second = int(second)

total_seconds1 = (hour * 3600) + (minute * 60) + second

print(f"เวลาทั้งหมด = {total_seconds1:,} วินาที")
````

ให้สร้างฟังก์ชันคำนวนเวลาเป็นวินาทีและปริ้นให้ออกมาให้สวยงาม

````python
def  total_seconds(Time):
  list_Time = Time.split(':')
  hour = int(list_Time[0])*3600
  minute = int(list_Time[1])*60
  second = int(list_Time[2])
  total_seconds = (hour + minute + second)
  print(f"เวลาทั้งหมด = {total_seconds:,} วินาที")
````

````python
total_seconds('12:30:15')
````
